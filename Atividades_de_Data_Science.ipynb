para x e [0, 1] para y.

16. Resolva a equacao diferencial de segunda ordem (d²y/dx²) +y = 0, com condicoes
iniciais y(0) = 1 e y′(0) = 0.

```python
from scipy.integrate import solve_ivp

# Redução de ordem: transformamos a EDO de 2ª ordem em um sistema de duas EDOs de 1ª ordem
# Seja y1 = y  e  y2 = y'
# dy1/dx = y2
# dy2/dx = -y1
def sistema(x, Y):
    y1, y2 = Y
    return [y2, -y1]

# Condições iniciais: y(0) = 1 e y'(0) = 0
y0 = [1, 0]
intervalo_x = (0, 10)
pontos_x = np.linspace(0, 10, 100)

# Resolução
solucao = solve_ivp(sistema, intervalo_x, y0, t_eval=pontos_x)

# solucao.y[0] é y(x) e solucao.y[1] é y'(x)
print("Primeiros 5 valores de X:", solucao.t[:5])
print("Primeiros 5 valores de y(x):", solucao.y[0][:5])
```

19. Realize a transformada de Fourier de um sinal discreto composto de senos
e cossenos.

```python
from scipy.fft import fft, fftfreq

fs = 100               # Frequência de amostragem (100 Hz)
t = np.linspace(0, 1, fs, endpoint=False) # 1 segundo de sinal

# Criação do sinal composto
f1, f2 = 5, 20
sinal = np.sin(2 * np.pi * f1 * t) + np.cos(2 * np.pi * f2 * t)

# Fourier (FFT)
espectro = fft(sinal)
frequencias = fftfreq(len(t), d=1/fs)

# Seleção das frequências positivas e cálculo da magnitude
indices_positivos = frequencias >= 0
freq_pos = frequencias[indices_positivos]
magnitudes = np.abs(espectro[indices_positivos]) * (2 / len(t))

print("Frequências (Hz):", freq_pos[:25])
print("Magnitudes:", np.round(magnitudes[:25], 2))
```

6. Plote dois graficos de linha em subplots diferentes: y = e^x e y = log(x).

```python
# Evita o 0 devio ao LOG
x = np.linspace(0.1, 5.0, 50)

# Calcula as funções
y1 = np.exp(x)  # e^x
y2 = np.log(x)  # log de x

# Cria a vizualização com 2 subplots empilhados (2 linhas, 1 coluna)
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(6, 8))

# Função exponencial
ax1.plot(x, y1, color = 'blue', label= '$e^x$')
ax1.set_title('Função Exponencial ($y = e^x$)')
ax1.set_xlabel('x')
ax1.set_ylabel('y')
ax1.grid(True)

# Função Logarítmica
ax2.plot(x, y2, color='red', label='$log(x)$')
ax2.set_title('Função Logarítmica ($y = log(x)$)')
ax2.set_xlabel('x')
ax2.set_ylabel('y')
ax2.grid(True)


plt.tight_layout()
plt.show()
```

8. Plote um grafico de linha com intervalos de confianca (sombramento) para
a funcao y = sin(x).

```python
x = np.linspace(30, 60, 100)

y = np.sin(x)

limite_inferior = y - 0.2
limite_superior = y + 0.2

plt.plot(x, y, color='grey', label='Média')

# Adiciona o sombreamento
# 'alpha' controla a opacidade do sobreamento, vai de 0 a 1
plt.fill_between(x, limite_inferior, limite_superior, color='blue', alpha=0.2)
plt.title("Gráfico de linha da função y = sin(x) com sobreamento")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")


plt.show()
```

11. Crie um grafico de barras empilhadas para tres categorias com tres grupos
de valores.

```python
import numpy as np
import matplotlib.pyplot as plt

# Definindo os 3 grupos (eixo X)
grupos = ['Grupo 1', 'Grupo 2', 'Grupo 3']

# Valores das 3 categorias
cat_a = np.array([10, 15, 20])
cat_b = np.array([12, 18, 15])
cat_c = np.array([8, 12, 25])

fig, ax = plt.subplots(figsize=(8, 6))

# A categoria A vai na base normal
ax.bar(grupos, cat_a, label='Categoria A', color='darkblue')

# A categoria B começa onde A termina (bottom=cat_a), e a C vai em cima dela
ax.bar(grupos, cat_b, bottom=cat_a, label='Categoria B', color='blue')
ax.bar(grupos, cat_c, bottom=cat_a + cat_b, label='Categoria C', color='lightblue')

ax.set_ylabel('Valores Totais')
ax.set_xlabel('\n Grupos')
ax.set_title('Gráfico Empilhado (Matplotlib + Numpy)')
ax.legend()

plt.show()
```

12. Plote um grafico de linha com uma funcao quadratica e ajuste os limites
do eixo para focar nos valores x = -5 a x = 5.

```python
def funcao_quadratica(a, b, c, x):
    return a * (x**2) + b * x + c

# Definição do intervalo de x
x = np.linspace(-5, 5, 100)

# Definição da função
y = funcao_quadratica(2, -4, 1,x)



# Criação da figura e do plot
plt.figure(figsize=(8, 6))
plt.plot(x, y, color="purple", linewidth=2)

plt.title("Gráfico da Função Quadratica")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.grid(True, linestyle=":", alpha=0.5, color = 'black')

# Exibe gráfico
plt.show()
```
