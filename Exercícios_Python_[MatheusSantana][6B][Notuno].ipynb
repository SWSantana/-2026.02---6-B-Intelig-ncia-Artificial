18. Número Primo
o Crie um programa que verifique se um número fornecido é primo.

```python
import math
num  = int(input('Escolha o numero a ser verificado: '))

if num <= 1:
  print(f'O numero {num} não é primo')

if num == 2:
  print(f'O numero {num} é primo')

if num % 2 == 0 and num > 2:
  print(f'O numero {num} não é primo')

#Se o número não tiver nenhum divisor menor ou igual à sua própria raiz quadrada, ele é primo
limite_divisao = math.floor(math.sqrt(num)) + 1

cont = 3

primo = True

while cont <= limite_divisao:
  if num % cont == 0:
    primo = False
    break
  cont += 2

if primo:
  print(f'O numero {num} é primo')
else:
  print(f'O numero {num} não é primo')
```

21.Busca Binária
o Implemente a busca binária para encontrar um elemento em uma
lista ordenada.

```python
lista = [1, 2, 3, 5, 10, 8, 11, 13, 15, 16, 20]

lista.sort()

elemento_alvo = 4

inicio = 0
fim = len(lista) - 1

while inicio <= fim:
  # A "//" faz o resultado da divisão devolver valor exato
    meio = (inicio + fim) // 2
    if lista[meio] == elemento_alvo:
      print(f"Elemento encontrado no índice {meio}!")
      break # Interrompe o while imediatamente

    # Se o valor do meio for MAIOR que o alvo, o alvo está na esquerda
    elif lista[meio] > elemento_alvo:
        fim = meio - 1 # traz o fim para logo antes do meio
    # Se o valor do meio for MENOR que o alvo, o alvo está na direita
    else:
        inicio = meio + 1 # joga o início para logo após o meio
else:
  print(f'O elemento {elemento_alvo} não esta presente na lista')
```

22. Criptografia de Cifra de César
o Crie uma função para criptografar uma mensagem usando a cifra de
César.


```python
def criptografar_cesar(texto, chave):
    resultado = ""

    for caractere in texto:
        # Verifica se é uma letra maiúscula
        if caractere.isupper():
            # ord('A') descobre o código numérico da letra A (65)
            # % 26 garante que se passar de Z, volta para o A
            novo_codigo = (ord(caractere) - ord('A') + chave) % 26
            resultado += chr(novo_codigo + ord('A'))

        # Verifica se é uma letra minúscula
        elif caractere.islower():
            # ord('a') descobre o código numérico da letra a (97)
            novo_codigo = (ord(caractere) - ord('a') + chave) % 26
            resultado += chr(novo_codigo + ord('a'))

        else:
            resultado += caractere

    return resultado

mensagem = "Mensagem secreta"
chave_deslocamento = 3

mensagem_cifrada = criptografar_cesar(mensagem, chave_deslocamento)
print("Mensagem Criptografada:", mensagem_cifrada)
# Saída esperada: "Phqvdjhp vhfuhwd"
```

```python
def descriptografar_cesar(texto_cifrado, chave):
    resultado = ""

    for caractere in texto_cifrado:
        # Verifica se é uma letra maiúscula
        if caractere.isupper():
            # Subtraí a chave para voltar as posições no alfabeto
            novo_codigo = (ord(caractere) - ord('A') - chave) % 26
            resultado += chr(novo_codigo + ord('A'))

        # Verifica se é uma letra minúscula
        elif caractere.islower():
            novo_codigo = (ord(caractere) - ord('a') - chave) % 26
            resultado += chr(novo_codigo + ord('a'))

        else:
            resultado += caractere

    return resultado

# --- Teste da Função ---
mensagem_cifrada = "Phqvdjhp vhfuhwd"
chave_deslocamento = 3

mensagem_original = descriptografar_cesar(mensagem_cifrada, chave_deslocamento)
print("Mensagem Descriptografada:", mensagem_original)
# Saída esperada: "Mensagem secreta"
```

24.Validação de CPF
o Implemente uma função para validar um CPF brasileiro.

```python
def valida_CPF(cpf: str) -> bool:
  cpf = "".join(caractere for caractere in cpf if caractere.isdigit())

  # 2. Validação básica de tamanho
  if len(cpf) != 11:
      return False

  # Elimina CPF com todos numeros iguais(contra a regra da receita)
  if cpf == cpf[0] * 11:
        return False

  # Primerio digito verificador
  soma_1 = 0
  multiplicador_1 = 10

  # Multiplica os 9 primeiros dígitos por pesos decrescentes de 10 a 2
  for i in range(9):
      soma_1 += int(cpf[i]) * multiplicador_1
      multiplicador_1 -= 1

  resto_1 = soma_1 % 11
  digito_1 = 0 if resto_1 < 2 else 11 - resto_1

  # Segundo digito verificador
  soma_2 = 0
  multiplicador_2 = 11

  # Multiplica os 10 primeiros dígitos (incluindo o primeiro verificador) por pesos de 11 a 2
  for i in range(10):
      soma_2 += int(cpf[i]) * multiplicador_2
      multiplicador_2 -= 1

  resto_2 = soma_2 % 11
  digito_2 = 0 if resto_2 < 2 else 11 - resto_2

  if int(cpf[9]) == digito_1 and int(cpf[10]) == digito_2:
      return True
  else:
      return False


cpf_teste_1 = "123.456.789-09"
cpf_teste_2 = "22222222222"

print(f"CPF 1 é válido? {valida_CPF(cpf_teste_1)}")
print(f"CPF 2 é válido? {valida_CPF(cpf_teste_2)}")
```

30.Jogo da Forca
o Desenvolva um jogo simples da forca onde o usuário deve adivinhar
uma palavra.

```python
palavra = "PYTHON"
letras_certas = []
erros = 0

print("--- JOGO DA FORCA ---")

while erros < 6:
    # Cria a exibição em apenas uma linha: mostra a letra se ela foi acertada, senão mostra '_'
    exibicao = [letra if letra in letras_certas else "_" for letra in palavra]
    print("\nPalavra:", " ".join(exibicao))
    print(f"Erros: {erros}/6")

    # Condição de vitória: se não houver mais nenhum "_", o jogador ganhou
    if "_" not in exibicao:
        print("\nVocê venceu! 🎉")
        break

    palpite = input("Digite uma letra: ").upper()

    # Validação rápida de erro ou acerto
    if palpite in palavra:
        if palpite not in letras_certas:
            letras_certas.append(palpite)
    else:
        print("Letra errada!")
        erros += 1
else:
    print(f"\nVocê perdeu! A palavra era: {palavra}")
```

31.Ordenação por Bolha
o Implemente o algoritmo de ordenação por bolha (Bubble Sort).

```python
def bubble_sort(lista):
    n = len(lista)

    # O laço externo controla quantas vezes vamos passar pela lista
    for i in range(n):
        # O laço interno faz as comparações par a par
        # O "- i - 1" é uma otimização: os últimos 'i' elementos já estão no lugar certo
        for j in range(0, n - i - 1):

            # Se o vizinho da esquerda for maior que o da direita, eles trocam
            if lista[j] > lista[j + 1]:
                # USADO NO EXERCICIO 10 TAMBEM
                lista[j], lista[j + 1] = lista[j + 1], lista[j]

    return lista


minha_lista = [64, 34, 25, 12, 22, 11, 90]
print("Lista original:", minha_lista)

lista_ordenada = bubble_sort(minha_lista)
print("Lista ordenada:", lista_ordenada)
```

```python
def desenhar_tabuleiro(tabuleiro):
    # Exibe o tabuleiro de forma organizada com linhas separadoras
    print("\n")
    print(f" {tabuleiro[0][0]} | {tabuleiro[0][1]} | {tabuleiro[0][2]} ")
    print("---|---|---")
    print(f" {tabuleiro[1][0]} | {tabuleiro[1][1]} | {tabuleiro[1][2]} ")
    print("---|---|---")
    print(f" {tabuleiro[2][0]} | {tabuleiro[2][1]} | {tabuleiro[2][2]} ")
    print("\n")

def verificar_vitoria(tab, j):
    # Verifica linhas, colunas e as duas diagonais para o jogador atual (j)
    for i in range(3):
        if (tab[i][0] == tab[i][1] == tab[i][2] == j) or \
           (tab[0][i] == tab[1][i] == tab[2][i] == j):
            return True

    if (tab[0][0] == tab[1][1] == tab[2][2] == j) or \
       (tab[0][2] == tab[1][1] == tab[2][0] == j):
        return True

    return False

# Inicio
# Preenchemos o tabuleiro com os números de 1 a 9 para facilitar a escolha das posições
tabuleiro = [
    ["1", "2", "3"],
    ["4", "5", "6"],
    ["7", "8", "9"]
]

jogador_atual = "X"
jogadas = 0

print("--- JOGO DA VELHA ---")

while jogadas < 9:
    desenhar_tabuleiro(tabuleiro)

    palpite = input(f"Jogador [{jogador_atual}], escolha uma posição livre (1-9): ")

    # Encontra a linha e a coluna correspondente ao número escolhido
    encontrou_posicao = False
    for l in range(3):
        for c in range(3):
            if tabuleiro[l][c] == palpite:
                tabuleiro[l][c] = jogador_atual
                encontrou_posicao = True
                break

    if not encontrou_posicao:
        print("Posição inválida ou já ocupada! Tente novamente.")
        continue

    jogadas += 1

    # Verifica se o jogador atual venceu após a jogada
    if verificar_vitoria(tabuleiro, jogador_atual):
        desenhar_tabuleiro(tabuleiro)
        print(f"Parabéns ao jogador [{jogador_atual}]")
        break

    # Alterna o jogador para a próxima rodada
    jogador_atual = "O" if jogador_atual == "X" else "X"
else:
    desenhar_tabuleiro(tabuleiro)
    print("Deu Velha")
```

36. Gerador de Senhas
o Crie um gerador de senhas aleatórias com letras e números.

```python
import secrets
import string

def gerar_senha(tamanho=12):
    # Junta todas as letras (maiúsculas e minúsculas) e números
    caracteres_disponiveis = string.ascii_letters + string.digits

    # Sorteia caracteres aleatórios seguros e os junta em uma única string
    senha = "".join(secrets.choice(caracteres_disponiveis) for _ in range(tamanho))

    return senha

print("--- GERADOR DE SENHAS ---")
tamanho_desejado = int(input("Digite o tamanho da senha (ex: 12): "))

senha_gerada = gerar_senha(tamanho_desejado)
print(f"Sua senha gerada é: {senha_gerada}")
```

37. Funções Recursivas
o Implemente uma função recursiva para calcular o fatorial de um
número

```python
def fatorial(n):
    # Se chegar 0 ou 1, retorna 1
    if n == 0 or n == 1:
        return 1

    #O número atual multiplica o fatorial do número anterior
    return n * fatorial(n - 1)

# --- Teste da Função ---
numero = 5
resultado = fatorial(numero)

print(f"O fatorial de {numero} é: {resultado}")
```

42. Números Primos até N
o Gere todos os números primos até um número n fornecido

```python
def eh_primo(num):
    # A mesma lógica do Exercicio 18
    if num <= 1:
        return False
    if num == 2:
        return True
    if num % 2 == 0:
        return False

    limite = math.floor(math.sqrt(num)) + 1
    for cont in range(3, limite, 2):
        if num % cont == 0:
            return False
    return True

n = int(input("Gerar números primos até: "))
lista_primos = []

# Varre todos os números de 2 até N (o n + 1 garante que N seja incluído no teste)
for numero in range(2, n + 1):
    if eh_primo(numero):
        lista_primos.append(numero)

print(f"Números primos até {n}:")
print(lista_primos)
```

46. Calcular Média e Desvio Padrão
o Calcule a média e o desvio padrão de uma lista de números.

```python
numeros = [10, 16, 23, 23, 16, 23, 21, 16]

n = len(numeros)

media = sum(numeros) / n

# Cálculo do Desvio Padrão
soma_variancia = sum((x - media) ** 2 for x in numeros)
variancia = soma_variancia / n

# A raiz quadrada da variância é o desvio padrão
desvio_padrao = math.sqrt(variancia)


print(f"Lista de números: {numeros}")
print(f"Média: {media:.2f}")
print(f"Desvio Padrão: {desvio_padrao:.2f}")
```

47. Contar Números em Lista

48.Ordenação por Inserção
o Implemente o algoritmo de ordenação por inserção (Insertion Sort).

```python
lista = [23, 10, 16, 21, 16, 23, 16]

# Comeca do segundo elemento (índice 1), pois o primeiro já está "ordenado" sozinho
for i in range(1, len(lista)):
    chave = lista[i]
    j = i - 1

    # Move os elementos da lista que são maiores que a chave
    # para uma posição à frente da sua posição atual
    while j >= 0 and lista[j] > chave:
        lista[j + 1] = lista[j]
        j -= 1
    lista[j + 1] = chave

print("Lista devidamente ordenada via Insertion Sort:")
print(lista)
```
